```python
from sklearn.datasets import make_blobs
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt

# Generate synthetic data
X, y = make_blobs(n_samples=500, n_features=5, centers=4, random_state=42)

# Apply t-SNE to reduce to 2 dimensions
tsne = TSNE(n_components=2, random_state=42, perplexity=40, max_iter=300)
X_embedded = tsne.fit_transform(X)

# Plot the reduced data
plt.figure(figsize=(8, 6))
for target, color in zip(set(y), ['red', 'green', 'blue', 'purple']):
    plt.scatter(
        X_embedded[y == target, 0],
        X_embedded[y == target, 1],
        color=color,
        label=f"Cluster {target}"
    )

plt.title("t-SNE on Synthetic Dataset")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.legend()
plt.grid()
plt.show()
```
